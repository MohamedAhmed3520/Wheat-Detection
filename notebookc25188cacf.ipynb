```python
import pandas as pd
import numpy as np
import ast
from sklearn.model_selection import train_test_split

# =========================
# 1. Load Data
# =========================
df = pd.read_csv("/kaggle/input/competitions/global-wheat-detection/train.csv")

# =========================
# 2. Split by image_id (IMPORTANT)
# =========================
image_ids = df['image_id'].unique()

train_ids, val_ids = train_test_split(
    image_ids,
    test_size=0.2,
    random_state=42
)

# =========================
# 3. Create train & val DataFrames (WITH COPY 🔥)
# =========================
train_df = df[df['image_id'].isin(train_ids)].copy()
val_df   = df[df['image_id'].isin(val_ids)].copy()

# =========================
# 4. Convert bbox string → list
# =========================
train_df['bbox'] = train_df['bbox'].apply(ast.literal_eval)
val_df['bbox']   = val_df['bbox'].apply(ast.literal_eval)

# =========================
# 5. Expand bbox → columns
# =========================
train_df[['xmin', 'ymin', 'width', 'height']] = pd.DataFrame(
    train_df['bbox'].tolist(), index=train_df.index
)

val_df[['xmin', 'ymin', 'width', 'height']] = pd.DataFrame(
    val_df['bbox'].tolist(), index=val_df.index
)

# =========================
# 6. Convert to xmax, ymax
# =========================
train_df['xmax'] = train_df['xmin'] + train_df['width']
train_df['ymax'] = train_df['ymin'] + train_df['height']

val_df['xmax'] = val_df['xmin'] + val_df['width']
val_df['ymax'] = val_df['ymin'] + val_df['height']

# =========================
# 7. Drop bbox column (optional)
# =========================
train_df.drop(columns=['bbox'], inplace=True)
val_df.drop(columns=['bbox'], inplace=True)

# =========================
# 8. Check
# =========================
print("Train shape:", train_df.shape)
print("Val shape:", val_df.shape)

print("\nSample Train:")
print(train_df.head())
```

```python
import os
import shutil
import cv2

# =========================
# PATHS
# =========================
SOURCE_IMG_DIR = "/kaggle/input/competitions/global-wheat-detection/train"   # مكان صور Kaggle الأصلية
BASE_DIR = "dataset"

# =========================
# CREATE FOLDERS
# =========================
for split in ["train", "valid"]:
    os.makedirs(f"{BASE_DIR}/{split}/images", exist_ok=True)
    os.makedirs(f"{BASE_DIR}/{split}/labels", exist_ok=True)

# =========================
# YOLO CONVERSION FUNCTION
# =========================
def convert_to_yolo(row, w, h):
    x_center = (row['xmin'] + row['xmax']) / 2 / w
    y_center = (row['ymin'] + row['ymax']) / 2 / h
    bw = (row['xmax'] - row['xmin']) / w
    bh = (row['ymax'] - row['ymin']) / h

    return f"0 {x_center:.6f} {y_center:.6f} {bw:.6f} {bh:.6f}"

# =========================
# FUNCTION TO PROCESS ANY DF
# =========================
def process_dataframe(df, split_name):
    for image_id, group in df.groupby('image_id'):
        img_path = os.path.join(SOURCE_IMG_DIR, image_id + ".jpg")

        img = cv2.imread(img_path)
        if img is None:
            continue

        h, w = img.shape[:2]

        # Convert all boxes
        yolo_lines = group.apply(lambda row: convert_to_yolo(row, w, h), axis=1)

        # Save labels
        label_path = f"{BASE_DIR}/{split_name}/labels/{image_id}.txt"
        with open(label_path, "w") as f:
            f.write("\n".join(yolo_lines) + "\n")

        # Copy image
        dest_img_path = f"{BASE_DIR}/{split_name}/images/{image_id}.jpg"
        shutil.copy(img_path, dest_img_path)

# =========================
# RUN ON BOTH
# =========================
process_dataframe(train_df, "train")
process_dataframe(val_df, "valid")

print("✅ train + valid جاهزين بالكامل")
```

```python
import os
import cv2
import shutil

# =========================
# PATHS
# =========================
TEST_IMG_DIR = "/kaggle/input/competitions/global-wheat-detection/test"
BASE_DIR = "/kaggle/working/dataset"

# =========================
# CREATE TEST FOLDERS
# =========================
os.makedirs(f"{BASE_DIR}/test/images", exist_ok=True)
os.makedirs(f"{BASE_DIR}/test/labels", exist_ok=True)

# =========================
# YOLO FORMAT FUNCTION
# =========================
def convert_to_yolo(row, w, h):
    x_center = (row['xmin'] + row['xmax']) / 2 / w
    y_center = (row['ymin'] + row['ymax']) / 2 / h
    bw = (row['xmax'] - row['xmin']) / w
    bh = (row['ymax'] - row['ymin']) / h

    return f"0 {x_center:.6f} {y_center:.6f} {bw:.6f} {bh:.6f}"

# =========================
# PROCESS TEST DF
# =========================
def process_test_dataframe(test):
    for image_id, group in test.groupby("image_id"):

        img_path = os.path.join(TEST_IMG_DIR, image_id + ".jpg")
        img = cv2.imread(img_path)

        if img is None:
            print(f"⚠️ Missing: {image_id}")
            continue

        h, w = img.shape[:2]

        # Convert boxes
        yolo_lines = group.apply(lambda row: convert_to_yolo(row, w, h), axis=1)

        # Save label
        label_path = f"{BASE_DIR}/test/labels/{image_id}.txt"
        with open(label_path, "w") as f:
            f.write("\n".join(yolo_lines) + "\n")

        # Copy image
        shutil.copy(img_path, f"{BASE_DIR}/test/images/{image_id}.jpg")

# =========================
# RUN
# =========================
process_test_dataframe(test)

print("✅ Test dataset جاهز (images + labels)")
```
